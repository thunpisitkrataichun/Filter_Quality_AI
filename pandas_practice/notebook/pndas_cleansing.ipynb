```python
import pandas as pd
from dotenv import load_dotenv
import os
```

```python
# https://www.youtube.com/watch?v=bDhvCp3_lYw
# กฏการ cleansing data :
# 1. เอา Record ที่ซ้ำออกให้หมดก่อน
# 2. ลบ Colume ที่ไม่จำเป็นออก
# 3. เช็คข้อมูลว่ามันถูกต้องหรือไม่
```

```python
load_dotenv()
myfilepath = os.getenv("FILEPATH")
dataframe = pd.read_excel(myfilepath)
dataframe
```

```python
dataframe.drop_duplicates()
dataframe = dataframe.drop_duplicates()
dataframe
```

```python
dataframe = dataframe.drop(columns="Not_Useful_Column")
```

```python
dataframe["Last_Name"]=dataframe["Last_Name"].str.strip("123._/") #จัดการกับเว้นวรรค

dataframe
```

```python
dataframe["Phone_Number"]=dataframe["Phone_Number"].str.replace('[^a-zA-Z0-9]','',regex=True)

dataframe
```

```python
dataframe["Phone_Number"] = dataframe["Phone_Number"].apply(lambda x : str(x))
dataframe["Phone_Number"] = dataframe["Phone_Number"].apply(lambda x : x[0:3] + '-' + x[3:6] + '-' +x[6:10])
dataframe["Phone_Number"] = dataframe["Phone_Number"].str.replace('nan--','') 
dataframe["Phone_Number"] = dataframe["Phone_Number"].str.replace('Na--','')
dataframe

```

```python
dataframe[["Street Address","State","Zip_code"]] =dataframe["Address"].str.split(',', n = 3 , expand=True)
dataframe
```

```python
dataframe = dataframe.drop(columns="Address")
dataframe
```

```python
dataframe["Paying Customer"] = dataframe["Paying Customer"].str.replace('NaN', '')
dataframe["Paying Customer"] = dataframe["Paying Customer"].str.replace('N/a', '')
dataframe["Paying Customer"] = dataframe["Paying Customer"].str.replace('Yes' , 'Y')
dataframe["Paying Customer"] = dataframe["Paying Customer"].str.replace('No','N')

dataframe["Do_Not_Contact"]  = dataframe["Do_Not_Contact"].str.replace('Yes' , 'Y')
dataframe["Do_Not_Contact"]  = dataframe["Do_Not_Contact"].str.replace('No','N')
dataframe = dataframe.fillna('')
```

```python
dataframe
```

```python
for record in dataframe.index:
    if dataframe.loc[record,"Do_Not_Contact"] == "Y" :
        dataframe.drop(record,inplace=True)
    
for record in dataframe.index:
    if dataframe.loc[record,"Phone_Number"] == "" :
        dataframe.drop(record,inplace=True)
        
dataframe
```

```python
dataframe = dataframe.reset_index(drop=True)
```

```python
dataframe
```

```python
dataframe.to_csv("clean_data.csv", index=True)
```
